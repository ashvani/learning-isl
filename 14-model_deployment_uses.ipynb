```python
import requests
import json

url = "http://localhost:8000/predict"  # Replace with your actual endpoint
data = {"title": "Fears for T N pension after talks",
        "description": "Unions representing workers at Turner   Newall say they are 'disappointed' after talks with stricken parent firm Federal Mogul."}
headers = {
    "Content-Type": "application/json"
}

try:
    response = requests.post(url, data=json.dumps(data), headers = headers)
    response.raise_for_status()  # Raise an exception for bad status codes (4xx or 5xx)
    print("Request successful!")
    print("Response content:", response.text)
except requests.exceptions.RequestException as e:
    print(f"Error sending request: {e}")
```
