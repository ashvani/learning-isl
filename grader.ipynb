```python
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score
import os
```

```python
def validate_output(df):
    columns = ['pred', 'Id']
    columns_missing = []
    for column in columns:
        if not column in df.columns:
            columns_missing.append(column)

    if len(columns_missing) > 0:
        raise  NameError(f'Your data should have {columns_missing} column(s)')

    if df['classe'].isna().sum() > 0:
        raise ValueError(f'The column `classe` can\'t have missing values')
```

```python
def generate_accuracy(df_pred, df_actual):
    print(df_pred.head())
    print(df_actual.head())
    df_pred = df_pred.merge(df_actual, how = 'left', on = 'Id') 
    score = accuracy_score(df_pred['classe'], df_pred['pred'])
    print(score)
    return score
```

```python
def generate_rank(folder_path, golden_output_path):
    df_actual = pd.read_csv(golden_output_path)
    files = os.listdir(folder_path)
    valid_files = []
    for file in files:
        filepath = os.path.join(folder_path, file)
        if os.path.isdir(filepath):
            print(f'Skipping {file} as it is a directory')
            continue
        else:
            root, ext = os.path.splitext(filepath)
            if ext.lower() == '.csv':
                valid_files.append(filepath)
            else:
                print(f'Skipping the file {file} as it is a non-csv file')

    roll = []
    name = []
    score = []
    for file in valid_files:
        basename = os.path.basename(file)
        root, ext = os.path.splitext(basename)
        roll_temp, name_temp = root.split("_")
        df_pred = pd.read_csv(file)
        try:
            validate_output(df_pred)
        except Exception as e:
            print(f'Unable to score the {file} file due to {e}')
            continue
        score_temp = generate_accuracy(df_pred, df_actual)
        roll.append(roll_temp)
        name.append(name_temp)
        score.append(score_temp)
```

```python
final_name = []
final_roll = []
final_score = []
df_actual = pd.read_csv(golden_output_path)
df_actual = df_actual[['Id', 'classe']]
for folder in dirs:
    path = os.path.join(root, folder)
    files = os.listdir(path)
    files = [file for file in files if file.endswith('.csv')]
    print(f'processing the folder {folder}')
    if len(files) > 0:
        file = os.path.join(path, files[0])
        df = pd.read_csv(file)
        print(df.head())
        # try:
        #     validate_output(df)
        # except Exception as e:
        #     print(f'Unable to score the {file} file due to {e}')
        #     continue
        splits = folder.split("_")
        name, roll = splits[0], splits[1]
        final_name.append(name)
        final_roll.append(roll)
        score = generate_accuracy(df, df_actual)
        final_score.append(score)
    print('\tDone!\n\n')
```
